```python
import pandas as pd
from google_play_scraper import reviews, Sort
import json
```

```python
app_id = 'com.spotify.music'
target_count = 10000  # mengambil 10.000 data

print(f"Target aplikasi: {app_id}")
print(f"Target jumlah data: {target_count} ulasan")
```

```python
try:
    result, continuation_token = reviews(
        app_id,
        lang='id',           # Bahasa Indonesia
        country='id',        # Negara Indonesia
        sort=Sort.NEWEST,    # Urutkan dari yang terbaru
        count=target_count,  # Jumlah data yang ingin diambil
        filter_score_with=None # Ambil semua skor
    )

    # Jika data yang didapat kurang dari target Kita akan ambil lagi dengan urutan 'MOST_RELEVANT'
    if len(result) < target_count:
        print(f"Baru terkumpul {len(result)} data. Mencoba mengambil lagi dengan urutan 'RELEVANT'...")
        remaining_needed = target_count - len(result)

        result_relevant, _ = reviews(
            app_id,
            lang='id',
            country='id',
            sort=Sort.MOST_RELEVANT,
            count=remaining_needed,
            filter_score_with=None
        )
        # Gabungkan hasil
        result.extend(result_relevant)

    print(f"Scraping selesai. Total data yang berhasil didapat: {len(result)}")

except Exception as e:
    print(f"Terjadi error saat scraping: {e}")
```

```python
df = pd.DataFrame(result)
```

```python
df_raw = df[['content', 'score', 'at']].copy()
df_raw.rename(columns={'content': 'Teks', 'score': 'Bintang', 'at': 'Tanggal'}, inplace=True)
```

```python
nama_file = 'dataset_spotify_RAW.csv'
df_raw.to_csv(nama_file, index=False, encoding='utf-8')

print(f"\nDataset mentah berhasil disimpan ke file: {nama_file}")
print(f"Jumlah total data yang tersimpan: {len(df_raw)}")
```
